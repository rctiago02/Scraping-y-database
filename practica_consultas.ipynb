```python
import sqlite3, pandas as pd
def run(sql, params=()):
    conn = sqlite3.connect("libros.db")
    display(pd.read_sql_query(sql, conn, params=params))
    conn.close()
```

```python
# 1. Traé todos los productos.
query1 = """
SELECT *
FROM books
"""
#run(query1)


# 2 Mostrá nombre y precio de todos.
query2 = """
SELECT title, price
FROM books
"""
#run(query2)


# 3. Mostrá solo los productos con precio mayor a 20.
query3 = """
SELECT *
FROM books
WHERE price > 20;
"""
#run(query3)




# 4. Ordená los productos por precio de menor a mayor.
query4 = """
SELECT *
FROM books
ORDER BY price;
"""
#run(query4)




# 5. Mostrá productos cuyo stock sea 0.

query5 = """
SELECT * 
FROM books
WHERE stock = 0;
"""
#run(query5)



# 6. Mostrá los productos de la categoría 3.

query6 = """
SELECT * 
FROM books
WHERE category_id = 3;
"""
#run(query6)


# 7. Buscá productos cuyo nombre contenga la palabra "libro". (pista: LIKE)

query7 = """
SELECT *
FROM books
WHERE title LIKE '% book %';
"""
run(query7)
```

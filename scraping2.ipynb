```python
from bs4 import BeautifulSoup
import requests
from urllib.parse import urljoin
import json
import os
import time

# Cambio por url de pagina 1 para despues ir cambiando a cada pagina
url = "https://books.toscrape.com/catalogue/page-1.html"
items = []

rating_number = {"One":1,"Two":2,"Three":3,"Four":4,"Five":5}

next_page = url



# mientras existan paginas siguientes, veridicado por boton next. otra forma de hacerlo es iterando
while next_page:

    html_page = requests.get(next_page)
    soup = BeautifulSoup(html_page.text, "lxml")
    books = soup.find_all("article", class_="product_pod")

    for b in books:
        # Title
        title = b.h3.a["title"]
        # Price
        price = b.find("p", class_="price_color").text
        # URL
        href = b.h3.a["href"]
        book_url = urljoin(next_page, href)


        detail = requests.get(book_url)
        soup_detail = BeautifulSoup(detail.text, "html.parser")

        # category
        category = soup_detail.select("ul.breadcrumb li a")[-1].get_text(strip=True)

        # Stock
        available = b.find("p", class_="instock availability").text.strip()
        if available == "In stock":
            in_stock = True
        else: 
            in_stock = False

        # Rating                   
        rating_tag = soup_detail.select_one("p.star-rating")
        classes = rating_tag.get("class", [])                        
        rating_word = classes[1]
        rating = rating_number[rating_word]                           


        # description
        description_tag = soup_detail.find("div", id="product_description")
        description = (
            description_tag.find_next_sibling('p').text.strip()
            if description_tag else None
        )

        items.append({"title": title, "category": category, "rating": rating, "URL": book_url, "price": price, "stock": in_stock, "description": description})

    time.sleep(2)

    # para cada pagina del 1 al 5
    botton_next = soup.find("li", class_="next")
    if botton_next:
        href_next = botton_next.a["href"]
        next_page = urljoin(next_page, href_next)
    else:
        break


# Guardar en JSON
with open('libros_scrapeados.json', 'w', encoding='utf-8') as f:
    json.dump(items, f, ensure_ascii=False, indent=2)

# Leer desde JSON
with open('libros_scrapeados.json', 'r', encoding='utf-8') as f:
    datos = json.load(f)
    print(f'Se guardaron {len(datos)} libros')
    print(datos[0])  # Mostrar el primero para validar estructura

# Abrir automáticamente el archivo (solo en Windows)

os.startfile('libros_scrapeados.json')

```

```python
from bs4 import BeautifulSoup
import requests
from urllib.parse import urljoin
import json, os, time, re, sys




# Cambio por url de pagina 1 para despues ir cambiando a cada pagina
url = "https://books.toscrape.com/catalogue/page-1.html"
items = []

rating_number = {"One":1,"Two":2,"Three":3,"Four":4,"Five":5}

next_page = url


# 🔹 ADICIÓN (función): resolver autores por título con Open Library
# Explicación:
# - Open Library es una API pública (sin API key) que permite buscar libros por título.
# - Usamos el endpoint /search.json con el parámetro "title".
# - Traemos hasta 1 mejor resultado y devolvemos la lista de autores (si los hay).
# - Incluye manejo de errores y timeouts para no romper el scraping si la API falla.
OPENLIB_SEARCH = "https://openlibrary.org/search.json"

def buscar_autores_por_titulo_openlib(titulo):
    """
    Devuelve una lista de autores (strings) usando Open Library por título.
    Estrategia simple:
    - Pedimos hasta 5 docs, priorizamos match exacto por 'title' (case-insensitive).
    - Si no hay match exacto, usamos el primer doc (que ya viene rankeado por relevancia).
    - Si la API falla o no hay datos, devolvemos [].
    """
    try:
        r = requests.get(OPENLIB_SEARCH, params={"title": titulo, "limit": 5}, timeout=15)
        r.raise_for_status()
        data = r.json()
    except requests.RequestException:
        return []

    docs = data.get("docs") or []
    if not docs:
        return []

    titulo_cf = titulo.casefold()
    exact = [d for d in docs if d.get("title", "").casefold() == titulo_cf]
    candidato = exact[0] if exact else docs[0]
    autores = candidato.get("author_name") or []

    # Normalizamos y removemos duplicados
    limpio, seen = [], set()
    for a in autores:
        a2 = a.strip()
        if a2 and a2.lower() not in seen:
            limpio.append(a2)
            seen.add(a2.lower())
    return limpio


# mientras existan paginas siguientes, veridicado por boton next. otra forma de hacerlo es iterando
while next_page:

    html_page = requests.get(next_page)
    soup = BeautifulSoup(html_page.text, "lxml")
    books = soup.find_all("article", class_="product_pod")

    for b in books:
        # Title
        title = b.h3.a["title"]

        # Price
        # 💥💥💥 
        txtprice = b.find("p", class_="price_color").text
        price = float(re.sub(r"[^\d.]", "", txtprice))

        # URL
        href = b.h3.a["href"]
        book_url = urljoin(next_page, href)


        detail = requests.get(book_url)
        soup_detail = BeautifulSoup(detail.text, "html.parser")

        # category
        category = soup_detail.select("ul.breadcrumb li a")[-1].get_text(strip=True)

        # Stock
        available = b.find("p", class_="instock availability").text.strip()
        if available == "In stock":
            in_stock = True
        else: 
            in_stock = False

        # Rating                   
        rating_tag = soup_detail.select_one("p.star-rating")
        classes = rating_tag.get("class", [])                        
        rating_word = classes[1]
        rating = rating_number[rating_word]                           


        # description
        description_tag = soup_detail.find("div", id="product_description")
        description = (
            description_tag.find_next_sibling('p').text.strip()
            if description_tag else None
        )

        # 🔹 ADICIÓN (llamar API): buscar autores por título
        # Explicación:
        # - Usamos la función buscar_autores_por_titulo_openlib(title).
        # - Retorna una lista (posiblemente vacía) con los nombres de los autores.
        # - Se agrega como campo "authors" en el item para persistirlo luego en JSON.
        authors = buscar_autores_por_titulo_openlib(title)
        # Pequeña pausa de cortesía para no saturar la API si hay muchos libros
        time.sleep(0.2)

        items.append({
            "title": title,
            "category": category,
            "rating": rating,
            "URL": book_url,
            "price": price,
            "stock": in_stock,
            "description": description,
            # 🔹 ADICIÓN (nuevo campo en el dataset): autores como lista
            # Explicación:
            # - Este campo nuevo te permitirá luego crear tablas 'autores' y 'libro_autor' (M:N) en tu DB.
            # - Mantenerlo como lista te conserva todos los coautores que reporte la API.
            "authors": authors
        })

    time.sleep(2)

    # para cada pagina del 1 al 5
    botton_next = soup.find("li", class_="next")
    if botton_next:
        href_next = botton_next.a["href"]
        next_page = urljoin(next_page, href_next)
    else:
        break


# Guardar en JSON
with open('libros_scrapeados.json', 'w', encoding='utf-8') as f:
    json.dump(items, f, ensure_ascii=False, indent=2)

# Leer desde JSON
with open('libros_scrapeados.json', 'r', encoding='utf-8') as f:
    datos = json.load(f)
    print(f'Se guardaron {len(datos)} libros')
    print(datos[0])  # Mostrar el primero para validar estructura

# Abrir automáticamente el archivo (solo en Windows)
os.startfile('libros_scrapeados.json')
```

```python
from bs4 import BeautifulSoup
import requests
from urllib.parse import urljoin
import json, os, time, re, sys
from random import uniform


# Cambio por url de pagina 1 para despues ir cambiando a cada pagina

BASE = "https://books.toscrape.com/"
url = urljoin(BASE, "catalogue/page-1.html")
next_page = url



rating_number = {"One":1,"Two":2,"Three":3,"Four":4,"Five":5}
items = []



# 🔹 ADICIÓN (función): resolver autores por título con Open Library
# Explicación:
# - Open Library es una API pública (sin API key) que permite buscar libros por título.
# - Usamos el endpoint /search.json con el parámetro "title".
# - Traemos hasta 1 mejor resultado y devolvemos la lista de autores (si los hay).
# - Incluye manejo de errores y timeouts para no romper el scraping si la API falla.
OPENLIB_SEARCH = "https://openlibrary.org/search.json"
GOOGLE_BOOKS = "https://www.googleapis.com/books/v1/volumes"

import requests

def get_author(title, category, lang="en", timeout=5):
    """
    Busca el autor de un libro usando primero Open Library y luego Google Books.

    Parámetros:
        - title (str): título del libro.
        - category (str, opcional): categoría o tema del libro.
        - lang (str): idioma preferido para Google Books (por defecto 'en').
        - timeout (int): tiempo máximo por request.

    Retorna:
        - str: nombre del autor o 'Desconocido'.
    """

    # 1️⃣ Buscar en Open Library
    try:
        openLibrary_resp = requests.get(
            "https://openlibrary.org/search.json",
            params={"title": title, "limit": 3},
            timeout=timeout
        )
        openLibrary_resp.raise_for_status()
        openLibrary_data = openLibrary_resp.json()
        docs = openLibrary_data.get("docs", [])
        for d in docs:
            autores = d.get("author_name")
            if autores:
                return autores[0].strip()


        # 2️⃣ Buscar en Google Books (fallback)
        query = f'intitle:"{title}"'
        query += f' subject:"{category}"'


        try:
            googlebooks_resp = requests.get(
                "https://www.googleapis.com/books/v1/volumes",
                params={
                    "q": query,
                    "maxResults": 1,
                    "orderBy": "relevance",
                    "langRestrict": lang
                },
                timeout=timeout
            )
            googlebooks_resp.raise_for_status()
            gb_data = googlebooks_resp.json()
            items = gb_data.get("items", [])
            if items:
                authors = items[0]["volumeInfo"].get("authors")
                if authors and len(authors) > 0:
                    return authors[0].strip()
        except:
            raise
    except requests.RequestException as e:
        print(f"[⚠️] Error al consultar APIs books: {e}")

    # 3️⃣ Si ninguna API devuelve autor
    return "Desconocido"


# 🔹 ADICIÓN: función para obtener sopa con reintentos
def get_soup(url_pagina_soup):
    for intento in range(3):  # 🔹 ADICIÓN: reintentos
        try:
            resp = requests.get(url_pagina_soup, timeout=20)
            resp.raise_for_status()
            return BeautifulSoup(resp.text, "lxml")
        except requests.RequestException:
            time.sleep(1.2 * (intento + 1))
    raise RuntimeError(f"No pude obtener {url_pagina_soup}")





# mientras existan paginas siguientes, veridicado por boton next. otra forma de hacerlo es iterando
while next_page:

    soup = get_soup(next_page)
    books = soup.find_all("article", class_="product_pod")

    for b in books:


        # Title
        title = b.h3.a["title"]


        # Price
        # 💥💥💥 
        txtprice = b.find("p", class_="price_color").get_text(strip=True)
        price = float(re.sub(r"[^\d.]", "", txtprice))


        # URL
        href = b.h3.a["href"]
        book_url = urljoin(next_page, href)


        # detalle 
        soup_detail = get_soup(book_url)


        # category
        breadcrumb_link = soup_detail.select("ul.breadcrumb li a")
        category = breadcrumb_link[-1].get_text(strip=True) if len(breadcrumb_link) >= 3 else "Unknown"


        # Rating                   
        rating_tag = soup_detail.select_one("p.star-rating")
        classes = rating_tag.get("class", []) if rating_tag else []                        
        rating_word = next((c for c in classes if c in rating_number), "One")
        rating = rating_number.get(rating_word, 1)                         


        # Stock
        available = soup_detail.find("p", class_="instock availability").text.strip()
        if available == "In stock":
            in_stock = True
        else: 
            in_stock = False


        # description
        description_tag = soup_detail.find("div", id="product_description")
        description = (
            description_tag.find_next_sibling('p').text.strip()
            if description_tag else None
        )


        # 🔹 ADICIÓN (llamar API): buscar autores por título
        # Explicación:
        # - Usamos la función get_autor(title, category).
        # - Retorna una lista [] con los nombres de los autores o Desconocido.
        # - Se agrega como campo "author" en el item para persistirlo luego en JSON.
        author = get_author(title, category)
        # Pequeña pausa de cortesía para no saturar la API si hay muchos libros
        time.sleep(0.15)

        items.append({
            "title": title,
            "category": category,
            "rating": rating,
            "URL": book_url,
            "price": price,
            "stock": in_stock,
            "description": description,
            # 🔹 ADICIÓN (nuevo campo en el dataset): autores como lista
            # Explicación:
            # - Este campo nuevo te permitirá luego crear tablas 'autores' y 'libro_autor' (M:N) en tu DB.
            # - Mantenerlo como lista te conserva todos los coautores que reporte la API.
            "authors": author
        })

    time.sleep(1)

    # para cada pagina del 1 al 5
    botton_next = soup.find("li", class_="next")
    if botton_next:
        href_next = botton_next.a["href"]
        next_page = urljoin(next_page, href_next)
        time.sleep(0.5)
    else:
        break


# Guardar en JSON
with open('libros_scrapeados.json', 'w', encoding='utf-8') as f:
    json.dump(items, f, ensure_ascii=False, indent=2)

# Leer desde JSON
with open('libros_scrapeados.json', 'r', encoding='utf-8') as f:
    datos = json.load(f)
    print(f'Se guardaron {len(datos)} libros')
    print(datos[0])  # Mostrar el primero para validar estructura

# Abrir automáticamente el archivo (solo en Windows)
os.startfile('libros_scrapeados.json')
```

```python
import sqlite3
import json
from pathlib import Path

DB_PATH = "books.db"
JSON_PATH = "libros_scrapeados.json"

DDL = """
PRAGMA foreign_keys = ON;

CREATE TABLE IF NOT EXISTS categories (
    id      INTEGER PRIMARY KEY,
    name    TEXT NOT NULL UNIQUE
);

CREATE TABLE IF NOT EXISTS authors (
    id      INTEGER PRIMARY KEY,
    name    TEXT NOT NULL UNIQUE
);

CREATE TABLE IF NOT EXISTS books (
    id          INTEGER PRIMARY KEY,
    title       TEXT NOT NULL,
    url         TEXT NOT NULL UNIQUE,
    price       REAL NOT NULL,
    rating      INTEGER NOT NULL CHECK (rating BETWEEN 1 AND 5),
    stock       INTEGER NOT NULL CHECK (stock IN (0,1)),
    description TEXT,
    category_id INTEGER NOT NULL,
    FOREIGN KEY (category_id) REFERENCES categories(id) ON DELETE RESTRICT
);

CREATE TABLE IF NOT EXISTS book_author (
    book_id   INTEGER NOT NULL,
    author_id INTEGER NOT NULL,
    PRIMARY KEY (book_id, author_id),
    FOREIGN KEY (book_id)   REFERENCES books(id)   ON DELETE CASCADE,
    FOREIGN KEY (author_id) REFERENCES authors(id) ON DELETE CASCADE
);

-- Índices recomendados
CREATE INDEX IF NOT EXISTS idx_books_category   ON books(category_id);
CREATE INDEX IF NOT EXISTS idx_books_rating     ON books(rating);
CREATE INDEX IF NOT EXISTS idx_books_price      ON books(price);
CREATE INDEX IF NOT EXISTS idx_authors_name     ON authors(name);
CREATE INDEX IF NOT EXISTS idx_categories_name  ON categories(name);
"""

def get_or_create(cursor, table, unique_col, value):
    """
    Inserta si no existe y devuelve el id de la fila cuyo `unique_col` == value.
    """
    cursor.execute(f"INSERT OR IGNORE INTO {table} ({unique_col}) VALUES (?)", (value,))
    cursor.execute(f"SELECT id FROM {table} WHERE {unique_col} = ?", (value,))
    row = cursor.fetchone()
    return row[0]

def normalize_authors(raw_authors):
    """
    El scraper puede guardar 'authors' como str o list.
    - Si es str vacío o 'Desconocido', lo guardamos igual como  ['Desconocido'] (puedes cambiar esta política).
    - Si es str con comas, lo partimos por coma y limpiamos espacios.
    - Si es list, lo normalizamos a lista de strings.
    """
    if raw_authors is None:
        return ["Desconocido"]

    if isinstance(raw_authors, list):
        out = []
        for a in raw_authors:
            if a is None:
                continue
            if isinstance(a, str):
                a = a.strip()
                if a:
                    out.append(a)
        return out or ["Desconocido"]

    if isinstance(raw_authors, str):
        s = raw_authors.strip()
        if not s:
            return ["Desconocido"]
        # Si vienen múltiples autores en un string, separar por coma
        if "," in s:
            parts = [p.strip() for p in s.split(",") if p.strip()]
            return parts or ["Desconocido"]
        return [s]

    # Cualquier otro tipo: fallback
    return ["Desconocido"]

def main():
    if not Path(JSON_PATH).exists():
        raise SystemExit(f"No se encontró {JSON_PATH}. Ejecutá primero tu scraper para generar el JSON.")

    with sqlite3.connect(DB_PATH) as conn, open(JSON_PATH, "r", encoding="utf-8") as f:
        conn.execute("PRAGMA foreign_keys = ON;")
        cur = conn.cursor()

        # Crear esquema
        for stmt in filter(None, DDL.split(";")):
            s = stmt.strip()
            if s:
                cur.execute(s + ";")

        data = json.load(f)

        # Inserción
        for i, item in enumerate(data, start=1):
            title       = item.get("title") or "Sin título"
            url         = item.get("URL")   or ""
            price       = float(item.get("price") or 0.0)
            rating      = int(item.get("rating") or 1)
            stock_bool  = bool(item.get("stock"))
            stock       = 1 if stock_bool else 0
            description = item.get("description")
            category    = item.get("category") or "Unknown"

            # categoría
            cat_id = get_or_create(cur, "categories", "name", category)

            # libro
            cur.execute("""
                INSERT OR IGNORE INTO books (title, url, price, rating, stock, description, category_id)
                VALUES (?, ?, ?, ?, ?, ?, ?)
            """, (title, url, price, rating, stock, description, cat_id))

            # obtener id libro (ya inserto o existente)
            cur.execute("SELECT id FROM books WHERE url = ?", (url,))
            book_id = cur.fetchone()[0]

            # autores (normalizar)
            authors = normalize_authors(item.get("authors"))

            for author_name in authors:
                auth_id = get_or_create(cur, "authors", "name", author_name)
                # relación M:N
                cur.execute("""
                    INSERT OR IGNORE INTO book_author (book_id, author_id)
                    VALUES (?, ?)
                """, (book_id, auth_id))

            # commit por lote pequeño (opcional)
            if i % 50 == 0:
                conn.commit()

        conn.commit()

    print("✅ Base de datos creada y datos insertados en", DB_PATH)

    # Ejemplos de verificación rápida
    with sqlite3.connect(DB_PATH) as c:
        c.row_factory = sqlite3.Row
        r = c.execute("SELECT COUNT(*) AS n FROM books").fetchone()
        print("Libros:", r["n"])
        r = c.execute("SELECT COUNT(*) AS n FROM authors").fetchone()
        print("Autores:", r["n"])
        r = c.execute("SELECT COUNT(*) AS n FROM categories").fetchone()
        print("Categorías:", r["n"])
        r = c.execute("SELECT COUNT(*) AS n FROM book_author").fetchone()
        print("Relaciones libro↔autor:", r["n"])

if __name__ == "__main__":
    main()
```

## Se importan beautifulsoup y requests

```python
from bs4 import BeautifulSoup
import requests
```

## Traemos con ``GET`` el contenido de la URL 

```python
import requests

url = "https://books.toscrape.com/"

# requests.get() es una función que manda una petición HTTP de tipo GET a la dirección que pongas en url.
html_page = requests.get(url)


print(html_page.status_code)   # 200 significa que todo salió bien
print(html_page.text[:500])    # mostramos solo los primeros 500 caracteres del HTML
```

## Parseamos == Separamos y ordenamos, el **contenido tipo texto** de la URL que trajimos

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_page.text, "html.parser")
print(soup.title.text)   # muestra el <title> de la página
```

### EJ de uso de beautifulsoup

```python

titulos = soup.find_all("h3")   # busca todos los <h3>
print(titulos[0])               # muestra la primera etiqueta <h3>
print(titulos[0].text)          # solo el texto dentro de esa etiqueta
```

## Agarramos los enlaces de la página.

```python

# buscamos todas las etiquetas <a>
links = soup.find_all("a")


for link in links[:10]:   # solo los primeros 5
    print(link.get("href"))
```

## Buscamos todas las ``etiquetas`` de article con el ``atributo product_pod``

![alt text](image.png)

```python
books = soup.find_all("article", class_="product_pod")
print(len(books))   # cuántos libros encontró
```

## Acceso a titulos de cada index
- ### 1. De cada ``article``, vamos al elemento ``h3/etiqueta<h3>``
- ### 2. De cada elemento ``h3`` vamos a al elemento ``a/etiqueta<a>``
- ### 3. De cada elemento ``a`` accedemos solo al atributo ``title``

```python
titles = []

for book in books:
    title = book.h3.a["title"]
    titles.append(title)
    print(title)

print("\n", titles)
```

## En una lista de diccionario vamos agregando cada titulo con su href

```python
items = []
for b in books:
    title = b.h3.a["title"]
    href = b.h3.a["href"]           # link relativo
    items.append({"title": title, "href": href})

print(items[0])
```

## Precios de libros

```python
prices = []
for b in books:
    price = b.find("p", class_="price_color").text
    prices.append(price)

print(prices)
```

## En una lista items: Titulo, href, precio y disponibilidad

```python
items = []
for b in books:
    title = b.h3.a["title"]
    href = b.h3.a["href"]           # link relativo
    price = b.find("p", class_="price_color").text


    available = b.find("p", class_="instock availability").text.strip()
    if available == "In stock":
        in_stock = True
    else: 
        in_stock = False

    items.append({"title": title, "href": href, "price": price, "stock": in_stock})

print(items)
```
